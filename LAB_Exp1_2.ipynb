```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv('/content/CAR DETAILS FROM CAR DEKHO.csv')
df.info()

print("\nMissing Values:")
print(df.isnull().sum())
df.describe()

print("\nMissing Value Percentage:")
print((df.isnull().sum()/len(df))*100)

sns.histplot(data = df, x="selling_price", kde=True)
plt.title("\nDistribution of Selling Price")
plt.xlabel("\nSelling Price")
plt.ylabel("\nFrequency")
plt.show()

sns.histplot(data = df, x="year", kde=True)
plt.title("\nDistribution of Manufacturing Year")
plt.xlabel("\nYear")
plt.ylabel("\nFrequency")
plt.show()
```
